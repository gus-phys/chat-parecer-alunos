# Extração do texto de documentos

```python
# Leitura do PDF com exemplos de pareceres

caminho_dos_exemplos = "../data/raw/exemplo_de_pareceres.pdf"

leitor = PdfReader(caminho_dos_exemplos)

len(leitor.pages)
```

```python
paginas = []
for i, p in enumerate(leitor.pages, start=1):
    text = p.extract_text() or ""
    paginas.append({"page": i, "text": text})

print(paginas[0]['text'][:1000])
```
